# Model Discovery Agents Part 1

As discussed, the idea here is to understand how well a modern system can do at discovering knowledge. Eventually such systems could be used to probe real-world experiments, but I don't think we can start there. 

To get started, I decided to use Kevin Murphy's framework proposed in the recent paper... I do not intend this to restrict us to only studying a system in this way, it's just a helpful starting point.

The basic setup needs a few different components which will be put together here: 

1. a "World Model" -- this is the simulator, which the agent/ discovery system can only interact with by choosing experiments and observing results. In a real-world scientific discovery setting this would also be an experimental system. 

2. a "Hypothesis Generator" -- this is the system doing the reasoning. On a super high level, you can imagine this is a system that has access to a bunch of context, and understands how to formulate a logical causal hypothesis. This can be a restricted system that enumerates and filters structual causal models, an LLM prompted with a bunch of context and making use of knowledge from pre and post-training, or of course a human using their years of training and reasoning ability. 

3. an "Experimental Design" module -- this effectively takes what the world model or experimental system allows it in terms of actions, looks at its hypothesis set, and chooses which intervention to perform. It can be rationally and interpretable in design, such as a Value of Information measure, or rather can be something proposed by an LLM-based agent or a human because it/they decide this is likely to be useful. 

4. a "Dataset" -- this is just a record of all experiments and observations performed thus far on the system. It can be literally just measurements, but it can also be all the scientific context too. 

5. a "Predictor" -- this is probing whether the hypothesis/ current belief is valid by testing held-out scenarios. 

In practice, some of these components can be combined, whereas some are strictly separate. For instance, we might call the Hypothesis Generation and Experimental Design module collectively the Experimenter. This could be an agentic system, it could be a human.
